### Aufgabe 2: Einen Merge-Konflikt erkennen und den Vorgang abbrechen

<details>
<summary>Mögliche Lösung und Beobachtungen anzeigen</summary>

```text
<<<<<<< HEAD
Jede Person kann ihre Zuständigkeit für eine gemeinsame Aufgabe selbst festlegen.
=======
Das Lernteam legt die Zuständigkeit für eine gemeinsame Aufgabe gemeinsam fest.
>>>>>>> origin/main
```

Nach der Lösung zeigt der vorgemerkte Diff die Änderung gegenüber Sams neuer Grundlage. `git rebase --continue` erstellt den neu aufgebauten ersten Beitrag. Der zweite Beitrag fügt eine andere Datei hinzu und wird ohne weiteren Konflikt neu aufgebaut.

Beide eigenen Commits bekommen neue Kennungen. Sams Commit behält seine Kennung. Ihr `main` bleibt beim dritten Commit und das Teamrepository bleibt unverändert.

`git rebase --skip` wäre hier keine Lösung. Es würde den gerade erneut anzuwendenden Regelbeitrag überspringen.

<details>
<summary>Mögliche Lösung und Beobachtungen anzeigen</summary>

| Beobachtung | Merge | Rebase |
|---|---|---|
| Gewünschte Regel | Vorschlag und gemeinsame Bestätigung | Dieselbe fachliche Regel |
| Ursprüngliche Arbeitscommits | Beide Kennungen bleiben erreichbar | Beide werden auf dem Arbeitsbranch durch neue Kennungen ersetzt |
| Zusätzliche Commits gegenüber Sams Stand | Zwei Arbeitscommits und ein Merge-Commit | Zwei neu aufgebaute Arbeitscommits |
| Ort der Konfliktlösung | Neuer Merge-Commit | Neu aufgebauter erster Arbeitscommit |
| Eigener Hauptzweig | Dritter Startcommit | Dritter Startcommit |
| Teamhauptzweig | Sams vierter Commit | Sams vierter Commit |

Merge bewahrt die ursprüngliche Entwicklung und hält die Zusammenführung eigens fest. Rebase ist hier eine passende Alternative, wenn das Team die noch nicht geteilten Beiträge linear hinter Sams Stand sehen möchte.

Wenn Sam auf den ursprünglichen Beiträgen weiterarbeitet, ist deren gemeinsame Grundlage zusätzlich wichtig. Merge kann die bisherigen Kennungen erhalten. Ein Rebase bereits geteilter Beiträge würde eine Absprache über die neue Grundlage erfordern. Beide Verfahren brauchen unabhängig davon eine fachlich sinnvolle Konfliktlösung.

</details>

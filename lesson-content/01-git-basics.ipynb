
### Ihr Auftrag bei TaskFlow

TaskFlow ist eine kleine Aufgabenverwaltung für Lernteams. Mira betreut das Projekt. Sie möchte, dass neue Teammitglieder die wichtigsten Informationen direkt im Repository finden.

Sie legen dafür ein eigenes TaskFlow-Übungsrepository an. Zunächst entsteht eine Projektbeschreibung. Danach ergänzen Sie zusammengehörige Einstiegshilfen und zuletzt einen vollständigen Dokumentationsstand.

Die Texte dürfen eigene Formulierungen enthalten. Zu jeder Datei gibt es ein Minimalbeispiel. Achten Sie auf den Zweck der Datei und darauf, welche Änderungen gemeinsam in einen Commit gehören.

### Aufgabe 2: Ein neues TaskFlow-Repository beginnen

Mira bittet Sie zunächst um eine kurze Projektbeschreibung. Legen Sie im Home-Verzeichnis den Ordner `taskflow-dokumentation` an, wechseln Sie hinein und initialisieren Sie Git.

Erstellen Sie anschließend mit nano eine Datei `README.md`. Sie soll erklären, wofür TaskFlow gedacht ist. Ein mögliches Minimalbeispiel ist:

```text
# TaskFlow

TaskFlow unterstützt Lernteams bei der Aufgabenplanung.
```

Ihre Beschreibung darf anders formuliert sein. Prüfen Sie danach mit `git status` den Zustand der neuen Datei.

**Vorhersage:** Wurde die Datei durch das Speichern in nano bereits versioniert?


<details>
<summary>Lösung und Kontrolle anzeigen</summary>

```bash
cd ~
mkdir taskflow-dokumentation
cd taskflow-dokumentation
git init
nano README.md
git status
```

Sie sehen den versteckten Ordner `.git` mit `ls -la`. Die README steht nach dem Speichern unter `Untracked files`. Sie ist damit noch nicht versioniert.

Falls der Ordner bereits existiert, prüfen Sie mit `git status` und `git log --oneline`, ob darin ein früherer Übungsstand liegt. Überschreiben Sie vorhandene Arbeit nicht.

</details>


### Aufgabe 3: Eine Datei vormerken

Die Projektbeschreibung ist als erster abgeschlossener Arbeitsschritt bereit. Nehmen Sie ausschließlich `README.md` in die Staging Area auf. Kontrollieren Sie den Status und erstellen Sie anschließend einen Commit mit einer präzisen Nachricht. Eine mögliche Nachricht lautet:

```text
Projektbeschreibung für Lernteams anlegen
```

Prüfen Sie danach den Status. Erklären Sie in eigenen Worten, was `working tree clean` über die noch nicht vorhandenen Teamdokumente aussagt.


<details>
<summary>Lösung und Kontrolle anzeigen</summary>

```bash
git add README.md
git status
git commit -m "Projektbeschreibung für Lernteams anlegen"
git status
```

`git add README.md` bereitet nur diese Datei für den nächsten Commit vor. Ein sauberer Status bedeutet, dass Git keine offenen Änderungen sieht. Er sagt nichts darüber aus, ob die Dokumentation schon vollständig ist.

</details>


### Aufgabe 4: Zwei zusammengehörige Dateien gemeinsam vormerken

Mira möchte zwei Einstiegshilfen, die zusammen veröffentlicht werden sollen:

- `TEAM.md` beschreibt kurz, wie das Team neue Ideen bespricht.
- `BEITRAGEN.md` erklärt neuen Mitgliedern, dass sie kleine Änderungen verständlich dokumentieren sollen.

Erstellen Sie beide Dateien mit nano. Je Datei reichen eine Überschrift und ein aussagekräftiger Satz. Sie können sich an diesen Minimalbeispielen orientieren:

```text
# Zusammenarbeit

Neue Ideen werden zuerst im Team besprochen.
```

```text
# Mitwirken

Kleine Änderungen erhalten eine verständliche Beschreibung.
```

Beide Dateien gehören zu demselben Einstiegspaket. Nehmen Sie sie daher mit **einem** Aufruf von `git add` gemeinsam auf. Kontrollieren Sie vor dem Commit, ob beide Dateien vorbereitet sind. Wählen Sie eine passende Commit-Nachricht.


<details>
<summary>Lösung und Kontrolle anzeigen</summary>

```bash
nano TEAM.md
nano BEITRAGEN.md
git add TEAM.md BEITRAGEN.md
git status
git commit -m "Einstiegshilfen für das Team ergänzen"
```

Die beiden Dateinamen nach `git add` zeigen, welche Inhalte vorbereitet werden. Im Status müssen beide unter `Changes to be committed` stehen. Eine andere präzise Commit-Nachricht ist ebenfalls passend.

</details>


### Aufgabe 5: Einen vollständigen Dokumentationsstand übernehmen

Mira möchte jetzt einen zusammenhängenden Stand für neue Mitglieder fertigstellen. Führen Sie diese drei Änderungen aus:

1. Ergänzen Sie in `README.md` einen Verweis auf `TEAM.md` und `BEITRAGEN.md`.
2. Legen Sie `CHANGELOG.md` an. Die Datei soll eine Überschrift und einen kurzen Hinweis auf die neuen Einstiegshilfen enthalten.
3. Legen Sie `notizen.txt` an. Halten Sie darin fest, dass das Team später eine Sortierung nach Abgabedatum diskutieren möchte.

Die drei Änderungen bilden gemeinsam den aktuellen Dokumentationsstand. Nehmen Sie deshalb diesmal alle neuen und geänderten Dateien aus dem Projektordner mit `git add .` auf. Kontrollieren Sie den Status.

Erstellen Sie den Commit ohne `-m`. Verfassen Sie die Nachricht in nano. Die erste Zeile soll den Dokumentationsstand knapp benennen. Ein nachfolgender Absatz soll erklären, dass die Sortieridee noch nicht umgesetzt ist. Die nano-Hinweise am unteren Rand zeigen Speichern und Beenden.


<details>
<summary>Lösung und Kontrolle anzeigen</summary>

```bash
nano README.md
nano CHANGELOG.md
nano notizen.txt
git add .
git status
git commit
```

`git add .` berücksichtigt in diesem Projekt alle neuen, geänderten und gelöschten Dateien im aktuellen Verzeichnis sowie seinen Unterverzeichnissen. Es passt hier, weil alle drei Änderungen zu einem gemeinsamen Dokumentationsstand gehören.

Eine mögliche Commit-Nachricht im Editor lautet:

```text
Dokumentationsstand für neue Teammitglieder ergänzen

Die Sortierung nach Abgabedatum bleibt eine Idee zur späteren Besprechung.
```

In der Statusausgabe vor dem Commit erscheinen mindestens `README.md`, `CHANGELOG.md` und `notizen.txt` als vorbereitet. Nach dem Commit meldet `git status` einen sauberen Arbeitsstand.

</details>


### Aufgabe 6: Die Historie lesen und Hilfe finden

Mira möchte einen Überblick über Ihre Arbeit. Zeigen Sie die Historie zuerst kurz und danach ausführlich an.

1. Welcher Commit ist der neueste?
2. Wo finden Sie die längere Nachricht aus Aufgabe 5?
3. Welche Identität steht bei Ihren Commits?
4. Wie beenden Sie die ausführliche Ansicht?

Sie haben außerdem vergessen, wofür `-m` steht. Suchen Sie den Eintrag zunächst in der kurzen Hilfe und öffnen Sie danach die ausführliche Hilfe zu `git commit`.


<details>
<summary>Lösung und Antworten anzeigen</summary>

```bash
git log --oneline
git log
git commit -h
git help commit
```

`git log --oneline` zeigt Kurzkennung und erste Nachrichtenzeile. Der neueste Commit steht oben. Die ausführliche Ansicht enthält Autor oder Autorin, Zeitpunkt sowie weitere Absätze der Commit-Nachricht. Beenden Sie den Pager mit `q`.

`-m` übergibt eine Commit-Nachricht direkt im Befehl. Mehrere `-m`-Optionen ergeben getrennte Absätze derselben Nachricht. `git commit -h` ist die kurze Hilfe. `git help commit` öffnet die ausführliche Hilfe.

</details>

### Auftrag von Mira

Mira hat die offenen Änderungen in diesem Repository vorbereitet. Prüfen Sie zuerst den Zustand und die bisherige Historie. Untersuchen Sie danach die betroffenen Dateien.

Vervollständigen Sie den angefangenen Dokumentationsstand und speichern Sie ihn als nachvollziehbaren Commit. Erstellen Sie anschließend eine neue Datei `WILLKOMMEN.md` für neue Teammitglieder. Die Datei soll eine Überschrift und einen kurzen Willkommenshinweis enthalten. Speichern Sie diese zweite Änderung als eigenen Commit.

Wählen Sie die Zahl der Commits, die Dateiauswahl und die Nachrichten so, dass Mira die beiden Arbeitsschritte nachvollziehen kann. Prüfen Sie zum Schluss Status und Historie.


<details>
<summary>Mögliche Musterlösung anzeigen</summary>

Zunächst helfen eine Status- und eine Historienansicht beim Einordnen. Die vorhandenen Änderungen betreffen `README.md` und die neue Datei `CHANGELOG.md`. Beide gehören zum vorbereiteten Dokumentationsstand.

Eine mögliche Umsetzung ist:

```bash
git status
git log --oneline
cat README.md
cat CHANGELOG.md
git add README.md CHANGELOG.md
git commit -m "Dokumentationshinweise vervollständigen"
nano WILLKOMMEN.md
git add WILLKOMMEN.md
git commit -m "Willkommenshinweis ergänzen"
git status
git log --oneline
```

`WILLKOMMEN.md` kann zum Beispiel eine Überschrift `# Willkommen bei TaskFlow` und einen Hinweis enthalten, zuerst README und TEAM.md zu lesen.

Ein weiterer sinnvoller Ablauf ist möglich. Entscheidend sind diese Kriterien:

- Die beiden schon offenen Dokumentationsänderungen werden gemeinsam gespeichert.
- `WILLKOMMEN.md` erhält einen eigenen nachvollziehbaren Commit.
- Die Commit-Nachrichten benennen die jeweilige Änderung.
- Der Status ist am Ende sauber.

</details>

### Ihr Auftrag bei TaskFlow

Mira hat ein kleines Repository mit einer Projektübersicht, Bedienhinweisen, Teamregeln und einer Ideenliste vorbereitet. Sie sollen zunächst nachvollziehen, wie dieser Stand entstanden ist.

Danach ergänzen Sie die Dokumentation. Einige Änderungen sind fertig und andere sollen noch nicht gespeichert werden. Entscheiden Sie deshalb vor jeder Korrektur, ob sie die Staging Area, das Arbeitsverzeichnis oder einen Commit betrifft.


### Aufgabe 1: Die vorhandene Historie untersuchen

Prüfen Sie zuerst den aktuellen Zustand und zeigen Sie danach die kurze Commit-Historie an.

Beantworten Sie anhand der Ausgaben diese Fragen:

1. Wie viele Commits enthält das vorbereitete Repository?
2. Welcher Commit ist der neueste?
3. In welchem Commit wurde `BEDIENUNG.md` aufgenommen?
4. Welche Änderung enthält dieser Commit?
5. Welche Dateien gehören zum neuesten Commit?

Beschränken Sie für die dritte Frage die Historie auf `BEDIENUNG.md`. Untersuchen Sie anschließend den gefundenen Commit mit `git show`. Sie können seine Kurzkennung aus der Historie übernehmen. Für die letzte Frage genügt eine Statistik des neuesten Commits.


<details>
<summary>Lösung und Antworten anzeigen</summary>

```bash
git status
git log --oneline
git log --oneline -- BEDIENUNG.md
git show HEAD~2
git show --stat HEAD
```

Die Historie enthält vier Commits. Der neueste Commit heißt `Änderungsübersicht und Ideenliste anlegen`. `BEDIENUNG.md` wurde im Commit `Bedienhinweise ergänzen` aufgenommen. Dieser Commit liegt zwei Vorgänger vor `HEAD` und ist deshalb auch über `HEAD~2` erreichbar.

`git show HEAD~2` zeigt neben den Commit-Angaben die neu angelegte Datei und ihre Zeilen. Die Statistik für `HEAD` nennt `CHANGELOG.md` sowie `notizen.txt`.

Sie können bei `git show` statt `HEAD~2` auch die Kurzkennung aus Ihrer Ausgabe verwenden. Die Kennung wird bei jedem neu vorbereiteten Repository anders aussehen.

</details>


### Aufgabe 2: Die drei Vergleiche praktisch unterscheiden

Mira möchte die Bedienhinweise um einen Filter ergänzen. Öffnen Sie `BEDIENUNG.md` mit nano und ergänzen Sie einen Satz. Ein mögliches Minimalbeispiel ist:

```text
Offene Aufgaben können mit einem Filter angezeigt werden.
```

Prüfen Sie die Änderung mit `git diff BEDIENUNG.md`. Nehmen Sie die Datei danach in die Staging Area auf. Vergleichen Sie nun die Ausgaben von `git diff` und `git diff --staged`.

Öffnen Sie die Datei erneut und ergänzen Sie eine zweite noch nicht abgestimmte Idee. Ein mögliches Minimalbeispiel ist:

```text
Eine automatische Priorisierung ist geplant.
```

Nehmen Sie diese zweite Änderung nicht in die Staging Area auf. Führen Sie anschließend diese drei Vergleiche aus:

```bash
git diff
git diff --staged
git diff HEAD
```

Ordnen Sie vor dem Aufklappen der Lösung zu, welcher Vergleich nur den fertigen Filtersatz zeigt, welcher nur die spätere Idee zeigt und welcher beide Änderungen zeigt.


<details>
<summary>Lösung und Kontrolle anzeigen</summary>

```bash
nano BEDIENUNG.md
git diff BEDIENUNG.md
git add BEDIENUNG.md
git diff
git diff --staged
nano BEDIENUNG.md
git diff
git diff --staged
git diff HEAD
```

Nach dem ersten Speichern zeigt `git diff BEDIENUNG.md` den neuen Filtersatz. Nach `git add BEDIENUNG.md` ist die Ausgabe von `git diff` leer. `git diff --staged` zeigt den für den nächsten Commit vorbereiteten Filtersatz.

Nach dem zweiten Speichern liegen zwei Fassungen derselben Datei in Git:

- `git diff` zeigt nur die noch nicht vorgemerkte Idee zur Priorisierung.
- `git diff --staged` zeigt nur den vorgemerkten Filtersatz.
- `git diff HEAD` zeigt beide Änderungen gegenüber dem letzten Commit.

Bei einer eigenen Formulierung sehen die konkreten Zeilen anders aus. Die Zuordnung der Bereiche bleibt gleich.

</details>


### Aufgabe 3: Eine offene Änderung verwerfen und den letzten Commit verbessern

Mira bestätigt den Filtersatz. Die Idee zur automatischen Priorisierung ist noch zu unklar und soll verworfen werden.

Stellen Sie `BEDIENUNG.md` im Arbeitsverzeichnis aus der Staging Area wieder her. Prüfen Sie danach mit den passenden Diff-Ansichten, dass nur der Filtersatz für den Commit vorbereitet ist.

Erstellen Sie anschließend bewusst einen Commit mit der zu ungenauen Nachricht `Filter`. Zeigen Sie seine Kurzkennung an. Ersetzen Sie diesen letzten noch nicht geteilten Commit danach durch einen Commit mit der Nachricht `Filter für offene Aufgaben dokumentieren`. Vergleichen Sie die Kurzkennungen vor und nach der Korrektur.


<details>
<summary>Lösung und Kontrolle anzeigen</summary>

```bash
git restore BEDIENUNG.md
git diff
git diff --staged
git commit -m "Filter"
git log --oneline -1
git commit --amend -m "Filter für offene Aufgaben dokumentieren"
git log --oneline -1
git status
```

`git restore BEDIENUNG.md` übernimmt die vorgemerkte Fassung aus der Staging Area in das Arbeitsverzeichnis. Dadurch verschwindet nur die spätere nicht vorgemerkte Idee. Der Filtersatz bleibt vorbereitet.

Nach `--amend` steht die präzisere Nachricht an der Spitze der Historie. Die Kennung hat sich geändert, weil Git einen neuen Commit erstellt hat. Der Status ist am Ende sauber.

</details>


### Aufgabe 4: Eine versehentlich vorgemerkte Datei zurücknehmen

Mira bittet um eine Ergänzung in `TEAM.md`. Halten Sie dort fest, dass fertige Änderungen vor dem Commit geprüft werden. Ergänzen Sie gleichzeitig in `notizen.txt` eine weitere Idee, die noch nicht beschlossen ist.

Ein mögliches Minimalbeispiel für `TEAM.md` ist:

```text
Fertige Änderungen werden vor dem Commit geprüft.
```

Ein mögliches Minimalbeispiel für `notizen.txt` ist:

```text
Idee: Aufgaben automatisch nach Priorität anordnen.
```

Nehmen Sie zunächst mit `git add .` versehentlich beide Änderungen in die Staging Area auf. Korrigieren Sie die Auswahl so, dass nur `TEAM.md` für den nächsten Commit vorbereitet bleibt. Prüfen Sie dabei, ob der Inhalt von `notizen.txt` im Arbeitsverzeichnis erhalten geblieben ist.

Verwerfen Sie danach die offene Änderung an `notizen.txt`. Speichern Sie nur die fertige Teamregel in einem Commit mit einer passenden Nachricht.


<details>
<summary>Lösung und Kontrolle anzeigen</summary>

```bash
nano TEAM.md
nano notizen.txt
git add .
git status
git restore --staged notizen.txt
git status
git diff notizen.txt
git restore notizen.txt
git diff notizen.txt
git commit -m "Prüfung fertiger Änderungen dokumentieren"
git status
```

Nach `git add .` stehen beide Dateien unter `Changes to be committed`. `git restore --staged notizen.txt` entfernt nur die Notizdatei aus der Staging Area. `git diff notizen.txt` zeigt, dass die zusätzliche Idee weiterhin im Arbeitsverzeichnis steht.

Das zweite `git restore` verwirft diese nicht vorgemerkte Änderung. Der anschließende Commit enthält deshalb nur die Änderung an `TEAM.md`. Der Status ist danach sauber.

</details>


### Aufgabe 5: Einen fehlerhaften Commit nachvollziehbar aufheben

Legen Sie eine Datei `FUNKTIONEN.md` an und dokumentieren Sie darin versehentlich eine Funktion, die TaskFlow nicht besitzt:

```text
# Funktionen

TaskFlow versendet automatische Erinnerungen.
```

Nehmen Sie die Datei auf und erstellen Sie den Commit `Automatische Erinnerungen dokumentieren`. Prüfen Sie anschließend den Status sowie den neuesten Commit.

Mira weist Sie nun auf den fachlichen Fehler hin. Der Commit soll als Teil der Historie sichtbar bleiben. Erzeugen Sie deshalb mit `git revert` einen neuen Commit, der die gespeicherte Änderung aufhebt. Verwenden Sie `HEAD`, um den gerade erstellten Commit auszuwählen.

Git öffnet nano mit einer vorbereiteten Revert-Nachricht. Sie können diese Nachricht übernehmen und den Editor schließen. Untersuchen Sie danach die beiden neuesten Commits sowie die Änderung des Revert-Commits.


<details>
<summary>Lösung und Kontrolle anzeigen</summary>

```bash
nano FUNKTIONEN.md
git add FUNKTIONEN.md
git commit -m "Automatische Erinnerungen dokumentieren"
git status
git log --oneline -1
git revert HEAD
git log --oneline -2
git show HEAD
git status
```

Der erste Commit nimmt `FUNKTIONEN.md` in die Historie auf. `git revert HEAD` erstellt einen neuen Commit mit der Gegenänderung. Da der ursprüngliche Commit nur diese neue Datei enthielt, entfernt der Revert-Commit die Datei wieder.

Beide Commits bleiben in der Historie sichtbar. `git show HEAD` zeigt die Entfernung der Datei und der Status ist am Ende sauber.

</details>

### Auftrag von Mira

Untersuchen Sie zuerst den Status, die Historie und die betroffenen Änderungen. Bringen Sie das Repository danach in diesen Zustand:

1. Der letzte eigene Commit enthält eine sinnvolle Änderung, besitzt aber nur die Nachricht `Update`. Er wurde noch nicht geteilt. Ersetzen Sie seine Nachricht durch `Filterhinweise dokumentieren`, ohne dabei eine weitere offene Änderung in diesen Commit aufzunehmen.
2. Die interne Testnotiz in `README.md` soll verworfen werden.
3. Die vorbereitete Änderung an `CHANGELOG.md` ist fertig. Speichern Sie sie als eigenen Commit mit einer passenden Nachricht.
4. Ein älterer Commit behauptet, dass TaskFlow Aufgaben als PDF exportieren kann. Diese Funktion existiert nicht. Heben Sie den Commit durch einen neuen nachvollziehbaren Commit auf.
5. Prüfen Sie abschließend, dass der Arbeitsstand sauber ist und die Historie Ihre Korrekturen verständlich zeigt.

Begründen Sie für jede Korrektur, welcher Git-Bereich betroffen war und warum das gewählte Werkzeug dazu passt.


<details>
<summary>Mögliche Musterlösung anzeigen</summary>

Eine mögliche Reihenfolge ist:

```bash
git status
git log --oneline
git diff
git diff --staged
git restore --staged CHANGELOG.md
git commit --amend -m "Filterhinweise dokumentieren"
git restore README.md
git add CHANGELOG.md
git commit -m "Änderungsübersicht um Filterhinweis ergänzen"
git log --oneline -- FUNKTIONEN.md
git revert HEAD~2
git status
git log --oneline
git show HEAD
```

Die Änderung an `CHANGELOG.md` wird zuerst aus der Staging Area genommen. Sonst würde `--amend` sie unbeabsichtigt in den Ersatz für `Update` aufnehmen. Die Datei bleibt dabei im Arbeitsverzeichnis verändert.

Die Testnotiz in der README ist nicht vorgemerkt und kann aus der gespeicherten Fassung wiederhergestellt werden. Das Changelog wird anschließend als eigener Arbeitsschritt gespeichert.

Nach diesem neuen Changelog-Commit liegt der fehlerhafte PDF-Commit zwei Vorgänger hinter `HEAD`. `git revert HEAD~2` erstellt die Gegenänderung als neuen Commit. Ein anderer sinnvoller Ablauf ist möglich, wenn dieselben fachlichen Zustände erreicht werden.

</details>

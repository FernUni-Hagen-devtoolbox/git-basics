### Aufgabe 1: Einen Beitrag auf einem eigenen Branch vorbereiten

4. Prüfen Sie die neue Datei im Status. Nehmen Sie sie in die Staging Area auf und speichern Sie den Beitrag mit einer verständlichen Commit-Nachricht.
5. Zeigen Sie die Historie mit Branch-Verweisen an. Welche Branches zeigen jetzt auf unterschiedliche Commits?

Ein Minimalbeispiel für die Datei ist:

```text
# Aufgaben im Lernteam

Wir besprechen vor der Aufteilung, wer welche Aufgabe übernimmt.
```

Nanos Bedienhinweise stehen am unteren Fensterrand. Sie müssen die Tastenkürzel nicht auswendig lernen.

<details>
<summary>Mögliche Musterlösung anzeigen</summary>

```bash
git status
git branch
git log --oneline --decorate
git switch -c docs/lernteam
nano LERNTEAM.md
git status
git add LERNTEAM.md
git commit -m "Abstimmung im Lernteam beschreiben"
git log --oneline --decorate --all
```

<details>
<summary>Mögliche Musterlösung anzeigen</summary>

`main` ist ein Vorgänger Ihres Arbeitscommits. Git rückt den Verweis deshalb per Fast-Forward vor. Es entsteht kein zusätzlicher Commit. Beide Branches zeigen danach auf Ihren Beitrag. Das Löschen des Arbeitsbranchnamens entfernt diesen Commit nicht aus der Historie von `main`.

### Aufgabe 5: Dieselben Filterhinweise durch Rebase integrieren

Wechseln Sie zum noch unveränderten Vergleichsrepository:

```bash
cd ../rebase
git status
git switch docs/filter
git rebase main
```

Die vorbereiteten Änderungen betreffen unterschiedliche Dateien. Der Rebase sollte deshalb ohne Konflikt abschließen. Falls er anhält, untersuchen Sie den Status. Mit `git rebase --abort` können Sie den laufenden Vorgang abbrechen und zum Ausgangsstand zurückkehren. Die Konfliktlösung folgt in Abschnitt 5.

Zeigen Sie direkt nach dem Rebase den Graphen an:

```bash
git log --graph --oneline --decorate --all
```

Vergleichen Sie ihn mit Ihren Notizen aus Aufgabe 3:

2. Welcher Commit ist der Vorgänger des neuen ersten Filtercommits?

4. Warum zeigt `--all` weiterhin auch die ursprünglichen Filtercommits?

Wechseln Sie danach zu `main` und übernehmen Sie `docs/filter`. Beobachten Sie, warum dieser abschließende Schritt jetzt ein Fast-Forward ist. Prüfen Sie den Status.

<details>
<summary>Mögliche Musterlösung anzeigen</summary>

```bash
git log --graph --oneline --decorate --all
git switch main
git merge docs/filter
git status
git log --graph --oneline --decorate --all
```

Rebase hat Ihre beiden Änderungen nacheinander auf Sams `main`-Commit erneut angewendet. Die Filtercommits haben neue Kennungen. Der erste neue Filtercommit hat jetzt Sams Teamregel-Commit als Vorgänger. Der zweite baut auf dem neuen ersten Filtercommit auf.

`main` blieb während des Rebase bei Sams Beitrag stehen. Erst das abschließende `git merge docs/filter` rückt `main` per Fast-Forward zum neuen zweiten Filtercommit vor. Ein zusätzlicher Merge-Commit ist dafür nicht nötig.

Die ursprünglichen Filtercommits bleiben über `sam/weiterarbeit` erreichbar. Dieser Branch wurde durch Ihren Rebase nicht verändert. `--all` zeigt beide Folgen. Die Historie von `main` allein ist dagegen linear.

</details>

<details>
<summary>Beobachtungen und mögliche Einordnung anzeigen</summary>

Vor der Integration enthält `main..sam/weiterarbeit` im Merge-Repository einen Commit: Sams Beispielbeitrag. Ihre ursprünglichen Filtercommits gehören bereits zur Historie von `main`.

Im Rebase-Repository zeigt die Abfrage drei Commits: die beiden ursprünglichen Filtercommits und Sams Beispielbeitrag. Ihre Ersatzcommits enthalten bereits die Filterhinweise. Gleichwertige Änderungen machen die ursprünglichen Kennungen aber nicht zu Vorgängern der neuen `main`-Historie.

Nach der gezeigten Integration sind die Dateien in beiden Varianten sinnvoll kombiniert. Der Merge-Vergleich enthält sieben erreichbare Commits. Ihre Filter-Nachrichten erscheinen je einmal.

Der Rebase-Vergleich enthält acht erreichbare Commits. Ihre beiden Filter-Nachrichten erscheinen je zweimal: einmal mit den ursprünglichen Kennungen aus Sams Grundlage und einmal mit den Ersatzkennungen. Der anschließende Merge mit Sam hat wieder eine Verzweigung aufgenommen. Die zuvor lineare Historie ist damit nicht mehr linear.

Das Beispiel zeigt keinen Datenverlust. Es zeigt zusätzliche Historie durch die Verbindung ursprünglicher und ersetzter Commits. Wenn das Team eine gemeinsame lineare Folge behalten möchte, müsste es auch Sams darauf aufbauende Arbeit bewusst auf die neue Grundlage übertragen. Das erfordert Abstimmung. Diese Übertragung und der Austausch geteilter Branches werden hier noch nicht durchgeführt.

Merge erhält die ursprüngliche gemeinsame Grundlage. Rebase ist für noch nicht geteilte Arbeitscommits gut geeignet, wenn eine lineare Folge gewünscht ist. Bei bereits darauf aufbauender Arbeit müssen alle Beteiligten die veränderte Grundlage berücksichtigen.

</details>

Je nach Git-Version und Einrichtung kann zusätzlich `origin/HEAD -> origin/main` angezeigt werden. Dieser Verweis bezeichnet den Standardbranch der Quelle und ist kein eigener Arbeitsbranch.

<details>
<summary>Mögliche Musterlösung anzeigen</summary>

### Aufgabe 5: Die Verzweigung durch Merge integrieren

Versuchen Sie zunächst eine Aktualisierung, die nur ein Fast-Forward erlaubt:

```bash
git pull --ff-only origin main
```

Die Integration soll wegen der Verzweigung abgelehnt werden. Prüfen Sie den Status und die Historie. Wurde Ihr Arbeitsbranch dadurch auf einen anderen Commit gesetzt?

Übernehmen Sie nun Sams geholten Beitrag durch **Merge**. Prüfen Sie vor der Integration den aktuellen Branch und den sauberen Arbeitsstand. Falls Nano für die Merge-Nachricht geöffnet wird, können Sie den vorgeschlagenen Titel übernehmen.

Untersuchen Sie den fertigen Dateistand sowie:

```bash
git show --no-patch HEAD
git log --graph --oneline --decorate -4
```

- Welche beiden Vorgänger verbindet der neue Commit?
- Ist Ihr ursprünglicher Lernplan-Commit mit derselben Kennung erhalten geblieben?
- Sind die Lernplanung und Sams wöchentliche Besprechung jetzt enthalten?

Teilen Sie den gemeinsamen Stand durch einen normalen Push. Warum ist der Push jetzt ein Fast-Forward, obwohl Ihr neuer Commit ein Merge-Commit ist?

<details>
<summary>Mögliche Musterlösung anzeigen</summary>

```bash
git status
git branch
git merge origin/main
git show --no-patch HEAD
git log --graph --oneline --decorate -4
git status
git push origin main
git status
```

Der zuvor abgelehnte Pull hat den Arbeitsbranch nicht bewegt. Die geholten Teamdaten sind vorhanden.

Der neue Merge-Commit verbindet Ihren Lernplan-Commit und Sams Fortschrittscommit. Ihre ursprüngliche Lernplan-Kennung bleibt erhalten. `LERNPLAN.md` und die ergänzte `FORTSCHRITT.md` gehören nun zum gemeinsamen Stand. Es gibt acht von `main` aus erreichbare Commits.

Sams bisheriger Teamcommit ist ein Vorgänger des Merge-Commits. Der Teambranch kann deshalb beim Push per Fast-Forward zum Merge-Commit vorrücken. Nach dem Push sind Ihr `main` und sein bekannter Upstream wieder gleichauf.

Als gebündelter Ablauf wäre bei üblicher Konfiguration auch `git pull --no-rebase origin main` möglich gewesen. Wir haben hier die geholten Beiträge zuerst geprüft und den Merge dann separat ausgeführt.

</details>

### Aufgabe 6: Einen weiteren noch nicht geteilten Beitrag durch Rebase integrieren

Jetzt entsteht eine zweite unabhängige Ergänzung. Legen Sie mit `nano FRAGEN.md` eine kurze Hilfe für neue Teammitglieder an. Verwenden Sie eine Überschrift und mindestens eine Frage mit einer hilfreichen Antwort.

Ein Minimalbeispiel ist:

```text
# Fragen im Lernteam

Wo notieren wir eine offene Frage zu einer Aufgabe?
Wir schreiben sie zur Aufgabe und besprechen sie im nächsten Treffen.
```

Speichern Sie die Datei als Commit. Notieren Sie die Kurzkennung und die Nachricht. Teilen Sie diesen Commit noch nicht.

Lassen Sie Sam eine Ergänzung zur Sortierung der Aufgabenlisten veröffentlichen:

```bash
bash ~/scripts/simulate-taskflow-remotes.sh sortierung ..
```

Versuchen Sie einen normalen Push. Auch dieser wird abgelehnt. Holen und prüfen Sie den Teambeitrag:

```bash
git push origin main
git fetch origin
git log --oneline main..origin/main
git diff main origin/main
git status
```

Der Diff zwischen den beiden Branch-Ständen zeigt sowohl Sams neue Sortierregel als auch den Unterschied durch Ihre noch nicht geteilte FRAGEN-Datei. Minuszeilen bei `FRAGEN.md` bedeuten hier, dass Ihre Datei im verglichenen Teamstand noch fehlt. Der Vergleich verändert Ihre Datei nicht. Der Historienbereich zeigt dagegen nur die vom Teamstand aus erreichbaren Commits, die Ihrem Branch noch fehlen.

Wählen Sie diesmal **Rebase** und führen Sie Holen und Integration in einem Aufruf aus:

```bash
git pull --rebase origin main
```

Vergleichen Sie anschließend Kennungen, Nachrichten und Vorgänger mit Ihren Notizen. Ist Ihr neuer Commit mit derselben Kennung erhalten? Ist Sams veröffentlichter Commit unverändert? Wird für diesen Schritt ein zusätzlicher Merge-Commit benötigt?

Prüfen Sie die Inhalte und teilen Sie danach den Stand durch einen normalen Push.

Die vorgesehenen Änderungen betreffen verschiedene Dateien und lassen sich konfliktfrei integrieren. Sollte ein Merge oder Rebase wegen anderer Änderungen anhalten, zeigt `git status` den Zwischenstand. Mit `git merge --abort` beziehungsweise `git rebase --abort` können Sie einen noch laufenden Vorgang abbrechen. Die Konfliktlösung folgt in Abschnitt 5.

<details>
<summary>Mögliche Musterlösung und Beobachtungen anzeigen</summary>

Vor der Simulation können Sie den eigenen Beitrag so speichern:

```bash
nano FRAGEN.md
git status
git add FRAGEN.md
git commit -m "Umgang mit offenen Fragen beschreiben"
git log --oneline -1
```

Nach dem im Auftrag gezeigten Rebase prüfen und teilen Sie den Stand:

```bash
git log --graph --oneline --decorate -4
git show --no-patch HEAD
git status
git push origin main
git status
```

Ihr Fragencommit erhält eine neue Kennung. Seine Änderung und Nachricht bleiben erhalten, sein Vorgänger ist jetzt Sams Sortiercommit. Sams veröffentlichter Commit wird nicht ersetzt.

Für diese zweite Integration entsteht kein neuer Merge-Commit. Die beiden neuen Beiträge folgen hintereinander. Der frühere Merge aus Aufgabe 5 bleibt in der älteren Historie erhalten. Rebase macht hier also die neue Beitragsfolge linear, nicht nachträglich die gesamte Projekthistorie.

Nach der Integration gibt es zehn von `main` aus erreichbare Commits. Der normale Push kann vom Teamstand zu Ihrem Ersatzcommit vorrücken. Ein Force-Push ist nicht nötig, weil Sie nur Ihren noch nicht geteilten Fragencommit neu aufgesetzt haben.

</details>

```bash
git branch -vv
git branch -r
git log --graph --oneline --decorate --all -3
git status
```

### Aufgabe 8: Sams eigene Arbeitskopie zum Beitrag ansehen

Bis hierhin hat Ihr Push Sams Arbeitsdateien nicht automatisch aktualisiert. Wechseln Sie kurz in sein Arbeitsrepository und übernehmen Sie für diesen Schritt Sams Rolle:

Auf Ihrem `main` ist `BENACHRICHTIGUNGEN.md` nicht vorhanden. Die Datei gehört zum geteilten Arbeitsbranch, der noch nicht integriert wurde. Der Hauptzweig ist sauber und mit seinem Upstream gleichauf.
